# Formulación del problema

Se considera una cadena binaria de longitud $m = 100$, cuyos elementos pertenecen al conjunto $\{0, 1\}$. El objetivo es estimar el número esperado de unos en las cadenas.

Cada estado representa una buena cadena binaria de longitud $100$. El algoritmo inicia en el estado
$$X_0 = (0, 0, \dots, 0),$$
que contiene cien ceros y satisface la condición de validez.

En cada iteración, se selecciona aleatoriamente una posición $i \in \{1, \dots, 100\}$, con probabilidad $1/100$, y se propone invertir el valor de dicha posición:
$$X'_i = 1 - X_i.$$

Si $X_i = 0$, el cambio a uno solamente se acepta cuando sus posiciones vecinas no contienen unos. Para las posiciones interiores, esto requiere que
$$X_{i-1} = 0, \quad X_{i+1} = 0.$$

En las posiciones extremas, únicamente se verifica la vecina existente. Si $X_i = 1$, el cambio a cero siempre es válido. Cuando una propuesta no cumple las condiciones, el estado permanece sin cambios.

Se genera una cadena de Markov sobre el grafo de buenas cadenas. Para reducir la influencia del estado inicial, se descartan las primeras $30\,000$ iteraciones. Se realizan $300\,000$ iteraciones adicionales, en las cuales se registra el número de unos de cada cadena.

La función que se evalúa en cada iteración es
$$r(X_t) = \sum_{i=1}^{m} X_i(t),$$
donde $r(X_t)$ representa el número de unos de la cadena en el instante $t$.

La simulación se obtiene mediante:

$$E[r(X)] = \frac{1}{N} \sum_{t=1}^{N} r(X_t),$$

```python
import random   # Se generan números aleatoreos

# se define la función
def caminata_aleatorias(m=100, num_pasos=300000, burn_in=30000):

    # Se inicia una buena cadena (todos ceros)
    actual = [0] * m
    conteo_unos = []
    
    # Se ejecuta la caminata aleatiria
    for paso in range(num_pasos + burn_in):
        # Elegir una posición al azar para intentar cambiar
        i = random.randint(0, m - 1)
        
        # Se propone el nuevo valor (si es 0 pasa a 1, si es 1 pasa a 0)
        propuesta = 1 - actual[i]
        
        # Se verifica si la propuesta es una "buena cadena" (sin 1's adyacentes)
        es_valida = True
        if propuesta == 1:
            if (i > 0 and actual[i - 1] == 1) or (i < m - 1 and actual[i + 1] == 1):
                es_valida = False
                
        # Si es valido en el grafo, se mueve al nuevo grafo 
        if es_valida:
            actual[i] = propuesta
            
        # Se recolentan las muestras
        if paso >= burn_in:
            conteo_unos.append(sum(actual))
            
    # Se calcula el valor esperado de 1´s
    esperado_unos = sum(conteo_unos) / len(conteo_unos)
    return esperado_unos

m = 100 # Tamaño de la cadena
resultado = caminata_aleatorias(m=m, num_pasos=300000, burn_in=30000)

print(f"Longitud de la cadena (m): {m}\n")
print(f"Número esperado de 1's (Estimación por caminata aleatoria en el grafo): {resultado:.4f}")
```
