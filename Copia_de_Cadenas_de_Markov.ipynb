```python
import matplotlib.pyplot as plt

p = Matrix([[1, 0]])

dias = []
prob_soleado = []

for i in range(50):
    dias.append(i)
    prob_soleado.append(float(p[0,0]))
    p = p * P

plt.plot(dias, prob_soleado)
plt.axhline(y=1/3, linestyle='--')

plt.xlabel("Días")
plt.ylabel("Probabilidad de día soleado")
plt.title("Probabilidad de un día soleado a largo plazo")
plt.grid()

plt.show()
```

Cadenas de Markov Ejemplo 3

```python
from sympy import *
import matplotlib.pyplot as plt

init_printing(use_unicode=True)

P = Matrix([
    [0.80, 0.15, 0.05],
    [0.06, 0.90, 0.04],
    [0.04, 0.06, 0.90]
])

p = Matrix([[1, 0, 0]])

urbana = []
suburbana = []
rural = []
pasos = []

for i in range(50):
    pasos.append(i)

    urbana.append(float(p[0,0]))
    suburbana.append(float(p[0,1]))
    rural.append(float(p[0,2]))

    p = p * P

plt.plot(pasos, urbana, label="Urbana")
plt.plot(pasos, suburbana, label="Suburbana")
plt.plot(pasos, rural, label="Rural")

plt.xlabel("Número de transiciones")
plt.ylabel("Probabilidad")
plt.title("Comportamiento de la cadena de Markov")

plt.legend()
plt.grid()

plt.show()
```
