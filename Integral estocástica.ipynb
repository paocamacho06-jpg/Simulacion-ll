En el ejercicio 1 se demostro que:

$$\mathbb{E}\left[\int_0^t W_s^2\,ds\right] = \frac{t^2}{2}$$

$$\operatorname{Var}\left[\int_0^t W_s^2\,ds\right] = \frac{t^4}{3}$$

**Se sustituye $t = 1$**

Para la esperanza:

$$\mathbb{E}\left[\int_0^1 W_s^2\,ds\right] = \frac{1^2}{2}$$

$$\mathbb{E}\left[\int_0^1 W_s^2\,ds\right] = 0.5$$

Para la varianza:

$$\operatorname{Var}\left[\int_0^1 W_s^2\,ds\right] = \frac{1^4}{3}$$

$$\operatorname{Var}\left[\int_0^1 W_s^2\,ds\right] = 0.3333$$

```python
import numpy as np

M = 20000
n = 500
dt = 1/n

np.random.seed(42)

resultados = []

# Se simulan las trayectorias 
for k in range(M // 1000):

    # Incrementos del movimiento browniano
    dW = np.sqrt(dt) * np.random.standard_normal((1000, n))

    # Se construye el movimiento browniano
    W = np.concatenate(
        [np.zeros((1000, 1)), np.cumsum(dW, axis=1)],
        axis=1
    )

    # Se aproxima la integral mediante suma de Riemann
    I = dt * np.sum(W[:, :-1]**2, axis=1)

    resultados.extend(I)

# Se caclula la esperanza y varianza
esperanza = np.mean(resultados)
varianza = np.var(resultados, ddof=1)

print("Esperanza simulada:", esperanza)
print("Varianza simulada:", varianza)
```
